```python
# ============================================================
#                NLP TEXT ANALYZER - COLAB
# ============================================================

# -----------------------------
# 1. INSTALL REQUIRED LIBRARIES
# -----------------------------
!pip install -q nltk spacy vaderSentiment gradio

# Download NLTK resources
import nltk

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

# Download spaCy English model
!python -m spacy download en_core_web_sm -q


# -----------------------------
# 2. IMPORT LIBRARIES
# -----------------------------
import nltk
import spacy
import gradio as gr

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer


# -----------------------------
# 3. INITIALIZE NLP TOOLS
# -----------------------------
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

sentiment_analyzer = SentimentIntensityAnalyzer()

# Load spaCy model for NER
nlp = spacy.load("en_core_web_sm")


# -----------------------------
# 4. NLP ANALYSIS FUNCTION
# -----------------------------
def analyze_text(text):

    # Check empty input
    if not text or not text.strip():
        return (
            "Please enter some text.",
            "",
            "",
            "",
            "",
            ""
        )

    # ========================================================
    # 1. SENTENCE TOKENIZATION
    # ========================================================
    sentences = sent_tokenize(text)

    sentence_result = "\n".join(
        f"{i + 1}. {sentence}"
        for i, sentence in enumerate(sentences)
    )


    # ========================================================
    # 2. WORD TOKENIZATION
    # ========================================================
    words = word_tokenize(text)

    word_result = "\n".join(
        f"{i + 1}. {word}"
        for i, word in enumerate(words)
    )


    # ========================================================
    # 3. STEMMING
    # ========================================================
    stemming_result = []

    for word in words:

        if word.isalpha():

            stem = stemmer.stem(word)

            stemming_result.append(
                f"{word}  →  {stem}"
            )

    stemming_result = "\n".join(stemming_result)


    # ========================================================
    # 4. LEMMATIZATION
    # ========================================================
    lemmatization_result = []

    for word in words:

        if word.isalpha():

            lemma = lemmatizer.lemmatize(word)

            lemmatization_result.append(
                f"{word}  →  {lemma}"
            )

    lemmatization_result = "\n".join(lemmatization_result)


    # ========================================================
    # 5. NAMED ENTITY RECOGNITION (NER)
    # ========================================================
    doc = nlp(text)

    if doc.ents:

        ner_result = "\n".join(
            f"{entity.text}  →  {entity.label_}"
            for entity in doc.ents
        )

    else:

        ner_result = "No named entities found."


    # ========================================================
    # 6. SENTIMENT ANALYSIS
    # ========================================================
    scores = sentiment_analyzer.polarity_scores(text)

    compound = scores["compound"]

    if compound >= 0.05:
        sentiment = "Positive 😊"

    elif compound <= -0.05:
        sentiment = "Negative 😞"

    else:
        sentiment = "Neutral 😐"


    sentiment_result = f"""
Overall Sentiment: {sentiment}

Positive Score : {scores['pos']:.3f}
Negative Score : {scores['neg']:.3f}
Neutral Score  : {scores['neu']:.3f}
Compound Score : {scores['compound']:.3f}
"""


    # ========================================================
    # RETURN ALL RESULTS
    # ========================================================
    return (
        sentence_result,
        word_result,
        stemming_result,
        lemmatization_result,
        ner_result,
        sentiment_result
    )


# ============================================================
# 5. CREATE USER INTERFACE
# ============================================================

with gr.Blocks(
    theme=gr.themes.Soft()
) as demo:

    # Title
    gr.Markdown(
        """
        # 🧠 NLP Text Analyzer

        Enter text below and perform multiple Natural Language
        Processing operations.

        **Features:** Word Tokenization • Sentence Tokenization •
        Stemming • Lemmatization • NER • Sentiment Analysis
        """
    )


    # -----------------------------
    # INPUT
    # -----------------------------
    text_input = gr.Textbox(
        label="✍️ Enter Your Text",
        placeholder="Enter your text here...",
        lines=7
    )


    # -----------------------------
    # BUTTON
    # -----------------------------
    analyze_button = gr.Button(
        "🔍 Analyze Text",
        variant="primary"
    )


    gr.Markdown("## 📊 Analysis Results")


    # -----------------------------
    # TOKENIZATION
    # -----------------------------
    with gr.Row():

        with gr.Column():

            sentence_output = gr.Textbox(
                label="📝 Sentence Tokenization",
                lines=10
            )

        with gr.Column():

            word_output = gr.Textbox(
                label="🔤 Word Tokenization",
                lines=10
            )


    # -----------------------------
    # STEMMING / LEMMATIZATION
    # -----------------------------
    with gr.Row():

        with gr.Column():

            stemming_output = gr.Textbox(
                label="🌱 Stemming",
                lines=12
            )

        with gr.Column():

            lemmatization_output = gr.Textbox(
                label="📖 Lemmatization",
                lines=12
            )


    # -----------------------------
    # NER / SENTIMENT
    # -----------------------------
    with gr.Row():

        with gr.Column():

            ner_output = gr.Textbox(
                label="🏷️ Named Entity Recognition (NER)",
                lines=12
            )

        with gr.Column():

            sentiment_output = gr.Textbox(
                label="❤️ Sentiment Analysis",
                lines=12
            )


    # -----------------------------
    # CONNECT BUTTON TO FUNCTION
    # -----------------------------
    analyze_button.click(
        fn=analyze_text,

        inputs=text_input,

        outputs=[
            sentence_output,
            word_output,
            stemming_output,
            lemmatization_output,
            ner_output,
            sentiment_output
        ]
    )


# ============================================================
# 6. LAUNCH APPLICATION
# ============================================================

demo.launch(
    share=True
)
```
